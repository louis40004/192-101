## Mini Bank — One-Day Account Session   *(Fintech / core banking)*

You are writing the part of a bank app that processes **one customer's transactions for a day** and then prints an end-of-day report. The program reads a **starting balance**, then reads transactions one at a time until the customer enters `0` to finish.

### How each transaction works
- A **positive** number is a **deposit** — add it to the balance.
- A **negative** number is a **withdrawal** — e.g. `-300` means "take out 300".
- A withdrawal must be **rejected** if it would make the balance go below zero (the account may not be overdrawn). A rejected withdrawal does **not** change the balance.
- `0` ends the day (it is not a transaction).

### Your program must
1. use a `while` loop that keeps reading transactions until the customer enters `0`;
2. count the number of **deposits** and the total amount deposited;
3. count the number of **successful withdrawals** and the total amount withdrawn;
4. count the number of **rejected** withdrawals, and print `  Rejected: insufficient funds` when one happens;
5. keep the **balance** correct throughout, and print a final report exactly like the sample;
6. still print a correct report if the customer enters `0` immediately (no transactions).

### Sample run
```
Starting balance: 1000
Transaction (+ deposit, - withdraw, 0 to end): 500
Transaction (+ deposit, - withdraw, 0 to end): -2000
  Rejected: insufficient funds
Transaction (+ deposit, - withdraw, 0 to end): -300
Transaction (+ deposit, - withdraw, 0 to end): 0
---- Session report ----
Deposits: 1 totalling 500.0
Withdrawals: 1 totalling 300.0
Rejected withdrawals: 1
Final balance: 1200.0
```


**Tip:** build it in stages — first the loop that reads until `0`; then deposits; then withdrawals with the reject rule; then the report. Allowed tools only: variables, `input()`, `int()`/`float()`, arithmetic (`+ - * / // %`), `if`/`elif`/`else`, `for`/`while`, `print()`. No functions or lists needed.

````python
# Mini Bank — One-Day Account Session
# Type your OWN code here (AI-use Level 0: no AI assistants).
#
# Suggested steps:
#   1) read the starting balance
#   2) set up your counters and totals
#      (deposits, withdrawals, rejected, total_in, total_out)
#   3) read transactions in a while loop until the user enters 0
#   4) inside the loop: deposit (+) vs withdrawal (-), with the overdraft reject rule
#   5) after the loop: print the report exactly like the sample run

balance = float(input("Starting balance: "))

# Write the rest of your program below:


deposits = 0
total_deposited = 0.0
withdrawals = 0
total_withdrawn = 0.0
rejected = 0
transaction = float(input("Transaction (+ deposit, - withdraw, 0 to end): "))

while transaction != 0:
    if transaction > 0:
        balance = balance + transaction
        deposits = deposits + 1
        total_deposited = total_deposited + transaction

    elif transaction < 0:
        withdrawal = -transaction

        if withdrawal <= balance:
            balance = balance - withdrawal
            withdrawals = withdrawals + 1
            total_withdrawn = total_withdrawn + withdrawal
        else:
            rejected = rejected + 1
            print("  Rejected: insufficient funds")

    transaction = float(input("Transaction (+ deposit, - withdraw, 0 to end): "))

print("---- Session report ----")
print("Deposits:", deposits, "totalling", total_deposited)
print("Withdrawals:", withdrawals, "totalling", total_withdrawn)
print("Rejected withdrawals:", rejected)
print("Final balance:", balance)
````
