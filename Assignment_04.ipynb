## Main program

Use `FLIGHT = 'TG105'` and start with `booked = ['1A', '12C', '20F']`. Then repeat:

## Your program
Write **all** your code in the single code cell below. Before submitting, test **every** test case above yourself and check that your results match exactly.

````python
def is_valid_seat(seat, booked):
    # Write your code here
    seat = seat.strip().upper()

    if len(seat) not in (2, 3):
        return 'invalid format'

    if not seat[-1].isalpha() or not seat[:-1].isdigit():
        return 'invalid format'

    row = int(seat[:-1])

    if row < 1 or row > 30:
        return 'row out of range'

    if seat[-1] < 'A' or seat[-1] > 'F':
        return 'invalid letter'

    if seat in booked:
        return 'already booked'

    return 'OK'


def calculate_fare(seat, age, bags):
    # Write your code here
    base_fare = 2500
    row = int(seat[:-1])

    if row <= 5:
        base_fare = base_fare * 2.5

    if seat[-1] == 'A' or seat[-1] == 'F':
        base_fare = base_fare + 200

    if age < 2:
        fare = base_fare * 0.10
    elif age <= 11:
        fare = base_fare * 0.75
    else:
        fare = base_fare

    if bags > 1:
        bag_fee = (bags - 1) * 500
    else:
        bag_fee = 0

    fare = fare + bag_fee

    return round(fare, 2)


def create_ticket(name, flight, seat, fare):
    # Write your code here
    words = name.split()
    clean_name = ' '.join(words).title()

    row = int(seat[:-1])

    if row <= 5:
        seat_class = 'Business'
    else:
        seat_class = 'Economy'

    last_word = words[-1].upper()

    if len(last_word) >= 3:
        booking_name = last_word[:3]
    else:
        booking_name = last_word

    booking = booking_name + '-' + flight + '-' + seat

    ticket = (
        "============ E-TICKET ============\n"
        f"Passenger : {clean_name}\n"
        f"Flight    : {flight}  BKK -> CNX\n"
        f"Seat      : {seat} ({seat_class})\n"
        f"Fare      : {fare:,.2f} THB\n"
        f"Booking   : {booking}"
    )

    return ticket



# ----------------------------- MAIN PROGRAM -----------------------------
FLIGHT = 'TG105'
booked = ['1A', '12C', '20F']

# Write your main program here


ticket_count = 0
total_revenue = 0

while True:
    name = input("Passenger name (or done): ")

    if ''.join(name.split()).lower() == 'done':
        break

    while True:
        seat = input("Seat: ")
        result = is_valid_seat(seat, booked)

        if result == 'OK':
            break

        print("  x " + result)

    seat = seat.strip().upper()

    age = int(input("Age: "))
    bags = int(input("Bags: "))

    fare = calculate_fare(seat, age, bags)

    booked.append(seat)
    ticket_count = ticket_count + 1
    total_revenue = total_revenue + fare

    ticket = create_ticket(name, FLIGHT, seat, fare)
    print(ticket)


print("Tickets sold :", ticket_count)
print(f"Total revenue: {total_revenue:,.2f} THB")
print("Booked seats :", ', '.join(booked))
````
